```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Student_ID": range(1, 21),
    "Study_Hours": [
        1, 2, 1.5, 3, 2.5,
        4, 3.5, 5, 4.5, 6,
        5.5, 7, 6.5, 8, 7.5,
        9, 8.5, 10, 9.5, 11
    ],
    "Exam_Score": [
        35, 40, 38, 45, 43,
        50, 48, 58, 55, 62,
        60, 68, 65, 75, 72,
        82, 78, 88, 85, 92
    ]
}

df = pd.DataFrame(data)

df.to_csv("study_hours_exam_scores.csv", index=False)

print("Dataset created successfully!")
print("Total student records:", len(df))

df.head()
```

```python
print("Dataset information:")
df.info()

print("\nStatistical summary:")
print(df[["Study_Hours", "Exam_Score"]].describe())
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Exam_Score"],
    edgecolor="black"
)

plt.title("Relationship Between Study Hours and Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("study_hours_score_scatter_plot.png", dpi=300)
plt.show()
```

```python
correlation = df["Study_Hours"].corr(df["Exam_Score"])

print(f"Correlation coefficient: {correlation:.3f}")

if correlation > 0.5:
    print("There is a strong positive relationship.")
elif correlation > 0:
    print("There is a positive relationship.")
elif correlation < -0.5:
    print("There is a strong negative relationship.")
elif correlation < 0:
    print("There is a negative relationship.")
else:
    print("There is little or no linear relationship.")
```

```python
print("RELATIONSHIP ANALYSIS")
print("---------------------")

print("1. Study hours and exam scores generally increase together.")
print("2. Students with more study hours tend to have higher scores in this sample.")
print("3. The scatter plot shows a positive relationship.")
print(f"4. The calculated correlation is {correlation:.3f}.")

print("\nNote: Correlation does not prove causation.")
```
